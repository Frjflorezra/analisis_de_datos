```python
import pandas as pd
from IPython.display import display

from hf_dataset import HFDatasetLoader
from dataset_summarizer import DatasetSummarizer
```

```python
cargador = HFDatasetLoader(
    dataset_id="scikit-learn/iris",
    split="train"
)

df_original = cargador.cargar()

assert isinstance(df_original, pd.DataFrame)

print("Dimensiones originales:", df_original.shape)
display(df_original.head())
```

```python
# Renombrar la columna y guardar el resultado.
df_transformado = df_original.rename(
    columns={"Species": "especie"},
    errors="raise"
)

print("Columnas del DataFrame transformado:")
print(df_transformado.columns.tolist())

display(df_transformado.head())
```

```python
resumen = DatasetSummarizer(df_transformado)

ruta_resumen = resumen.generar_resumen("resumen_iris.txt")

print("Archivo generado en:")
print(ruta_resumen.resolve())
```

```python
contenido = ruta_resumen.read_text(encoding="utf-8")

# Verificar que la transformación se aplicó.
assert "especie" in df_transformado.columns
assert "Species" not in df_transformado.columns
assert df_transformado.shape == df_original.shape

# Verificar que el resumen contiene los datos transformados.
assert ruta_resumen.is_file()
assert "especie" in contenido
assert "Species" not in contenido
assert f"Cantidad de filas: {len(df_transformado)}" in contenido
assert (
    f"Cantidad de columnas: {df_transformado.shape[1]}"
    in contenido
)

print(contenido)
print("\n Flujo ejecutado correctamente.")
```

### Transformación realizada

Se cargó el dataset `scikit-learn/iris` desde Hugging Face y se
convirtió a un DataFrame de pandas.

La transformación consistió en renombrar la columna `Species`
a `especie`, conservando las filas y los valores originales.

El resultado se guardó en `df_transformado` y se pasó a la clase
`DatasetSummarizer`, que generó el archivo `resumen_iris.txt`.

El resumen incluye la cantidad de filas y columnas, los nombres
y tipos de columnas, los valores nulos y las estadísticas numéricas.
